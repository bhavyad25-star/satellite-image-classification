Satellite Image Classification for Earth Observation
This project explores machine-learning and deep-learning methods for classifying satellite images into land-cover categories using the EuroSAT dataset.
Objectives
- Explore and visualize satellite imagery.
- Build a traditional machine-learning baseline.
- Train a Convolutional Neural Network (CNN).
- Compare models using appropriate evaluation metrics.
- Allow users to upload their own images for prediction.
Tools: Python, NumPy, Matplotlib, scikit-learn, PyTorch, and Google Colab.

```python

# Step 1: Import the libraries we need

import os
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import sklearn

# Step 2: Check the software versions

print("NumPy version:", np.__version__)
print("PyTorch version:", torch.__version__)
print("Scikit-learn version:", sklearn.__version__)

# Step 3: Check whether a GPU is available

if torch.cuda.is_available():
    print("GPU available:", torch.cuda.get_device_name(0))
else:
    print("GPU not detected. We can still start with CPU.")

# Step 4: Make experiments reproducible

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

print("Environment setup completed!")
```

```python

# Import the dataset and visualization tools
from torchvision.datasets import EuroSAT
from torchvision import transforms
from torch.utils.data import random_split

# Download the EuroSAT RGB dataset
dataset = EuroSAT(
    root="./data",
    download=True
)

# Display basic dataset information
print("Total images:", len(dataset))
print("Number of classes:", len(dataset.classes))
print("Class names:", dataset.classes)

# Display the dimensions of one image
image, label = dataset[0]

print("Image size:", image.size)
print("Image label:", dataset.classes[label])

# Show the image
plt.figure(figsize=(5, 5))
plt.imshow(image)
plt.title(f"Class: {dataset.classes[label]}")
plt.axis("off")
plt.show()
```
